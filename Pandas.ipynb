```python
#data in dictionary --- key and value
data = {
    'Name':['Umar','Sherry','Nabeel','Usman'],
    'Age':['20','23','25','20'],
    'City':['Hyderabad','Lahore','Karachi','Lahore'],
    'Salary':['200','203','250','2000'],
}
```

```python
print(data)
```

```python
pd.DataFrame(data)
```

```python
# data in array  --- array in array
data_list = [
    ['Umer',20,'Lahore',300],
    ['Shehreyar',23,'Karachi',200],
    ['Nabeel',25,'Islamabad',250],
    ['Usman',20,'Lahore',400],
]
```

```python
# in this me label the array
df2= pd.DataFrame(data_list)
col = ['Name', 'Age', 'City', 'Salary']
df3 = pd.DataFrame(data_list,columns=col)
```

```python
df2
```

```python
df3
```

```python
# read specific column
df3[['Name','Age']]
```

```python
# adding column
df3['Designation']=['AI Eng','Ml Eng','Web Dev', ' DevOps']
```

```python
df3
```

```python
#removing the column
# axis= 1 (true) and axis = 0 (False)
df3.drop('Designation', axis=1)
```

```python
# Handling row we use location
df3.loc[[0,1]]
```

```python
df3.loc[[0,1]][['Salary', 'Designation']]
```

```python
# data handling by condition
df3[df3['Age']>22]
```

```python
df3[(df3['Age']>22) & (df3['City']=='Karachi')]
```

```python
df3[(df3['Age']>22) | (df3['City']=='Karachi')]
```
